### Ejercicio: Datos climáticos estación

Crea un nuevo DataFrame con datos de 1 estaciones SENAMHI de diferentes regiones del Perú (https://www.senamhi.gob.pe/site/descarga-datos/).

Los datos de las columnas corresponden a:

- Columna E: Temperatura máxima
- Columna F: Temperatura mínima

En las celdas donde aparece -99.9 significa que no hay información disponible para esa variable

1. Calcula el promedio anual de precipitación

```python
# Tu código aquí – Ejercicio: Estación

# Leer los datos de las estaciones y crear un DataFrame combinado
import os
files = os.listdir('/content/drive/MyDrive/2026-2/CIENCIA DE DATOS AMB/SENAMHI')  # Lista los archivos en el directorio
files
```

```python
# Parsear el nombre del archivo para obtener el código y nombre de la estación
def parsear_estacion(station):
    station = station.replace('.txt', '')
    partes = station.split('_', 1)
    codigo = partes[0]
    nombre = partes[1]
    return codigo, nombre

codigo, nombre= parsear_estacion(files[2])
print(f"Código: {codigo}, Nombre: {nombre}")
```

```python
# Leer el archivo CSV de la estación seleccionada
import pandas as pd

def leer_estacion(station):
  station_df = pd.read_table(f'/content/drive/MyDrive/2026-2/CIENCIA DE DATOS AMB/SENAMHI/{station}', header=None, sep=' ', na_values=['-999','-99.9']) # Sin encabezados
  station_df.columns = ['Año', 'Mes', 'Dia', 'Precipitación acumulada','Temperatura máxima', 'Temperatura mínima']  # Asignar nombres de columnas
  # Formatear las columnas
  station_df['Año'] = station_df['Año'].astype(int)
  return station_df

station_df = leer_estacion(files[2])
station_df.head()
```

```python
# Precipitacion mensual
precipitacion_mensual = station_df.groupby(['Año', 'Mes'])['Precipitación acumulada'].sum()

# Precipitacion total anual
precipitacion_anual = station_df.groupby('Año')['Precipitación acumulada'].sum()

# Dia mas lluvioso
dia_mas_lluvioso = station_df.loc[station_df['Precipitación acumulada'].idxmax()]

print(f"Precipitación mensual:\n{precipitacion_mensual}\n")
print(f"Precipitación anual:\n{precipitacion_anual}\n")
print(f"El día más lluvioso fue el {dia_mas_lluvioso['Año']:.0f}-{dia_mas_lluvioso['Mes']:.0f}-{dia_mas_lluvioso['Dia']:.0f} con {dia_mas_lluvioso['Precipitación acumulada']} mm de precipitación.")
```

```python
#Item 1
for i in files:
  codigo, nombre = parsear_estacion(i)
  print(f"Código: {codigo}, Nombre: {nombre}")

#Item 2
estaciones=[]
for i in files:
  codigo, nombres=parsear_estacion(i)

  station_df=leer_estacion(i)

  station_df['Fecha'] = pd.to_datetime(
    station_df[['Año', 'Mes', 'Dia']].rename(
        columns={'Año':'year', 'Mes':'month', 'Dia':'day'}))

  estacion = station_df[['Fecha', 'Precipitación acumulada']]

  estacion = estacion.rename(
    columns={'Precipitación acumulada': nombres})

  estaciones.append(estacion)

df_combinado = estaciones[0]

for estacion in estaciones[1:]:
    df_combinado = df_combinado.merge(estacion, on='Fecha', how='outer')

#Item 3 e Item 4
def analizar_estacion(station_df):
  # Promedio mensual
  precipitacion_mensual = station_df.groupby('Mes')['Precipitación acumulada'].mean()
  # Mes más lluvioso
  mes_mas_lluvioso = precipitacion_mensual.idxmax()
  return precipitacion_mensual, mes_mas_lluvioso


promedios={}
meses_lluviosos=[]
print()
print("-----------ESTACIONES------------")
for i in files:
    codigo, nombre = parsear_estacion(i)
    station_df = leer_estacion(i)

    precipitacion_mensual, mes_mas_lluvioso = analizar_estacion(station_df)

    promedios[nombre] = precipitacion_mensual
    meses_lluviosos.append([nombre, mes_mas_lluvioso])

    print(f"Estación: {nombre}")
    print(precipitacion_mensual)
    print(f"Mes más lluvioso: {mes_mas_lluvioso}")
    print()




#Item 5
promedio_mensual_df = pd.DataFrame(promedios)

mes_lluvioso_df = pd.DataFrame(
    meses_lluviosos,
    columns=['Estación', 'Mes más lluvioso']
)


promedio_mensual_df.to_csv(
    '/content/drive/MyDrive/2026-2/CIENCIA DE DATOS AMB/Tarea3_promedio_mensual_estaciones.csv'
)
```
